```python
def score_sucre(x):
    if pd.isnull(x):
        return x
    if x <= 4.5:
        return 0
    if x <= 9:
        return 1    
    if x <= 13.5:
        return 2 
    if x <= 18:
        return 3
    if x <= 22.5:
        return 4
    if x <= 27:
        return 5 
    if x <= 31:
        return 6
    if x <= 36:
        return 7
    if x <= 40:
        return 8
    if x <= 45:
        return 9
    return 10
```

```python
def score_gras(x):
    
    if pd.isnull(x):
        return x
    if x <= 10 :
        return int(x)
    return 10
```

```python
0.19//0.09
```

```python
def score_sel(x):
    
    if pd.isnull(x):
        return x
    if x <= 0.9:
        return x//0.09
    return 10       
```

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols

mod = ols('value ~ treatments', data=df_melt).fit()
                
aov_table = sm.stats.anova_lm(mod, typ=2)
print (aov_table)
```

```python
def Score_sante (x):
    if pd.isnull(x):
        return x
    if x <= 1/6:
        return 1
    if x <= 2/6:
        return 2    
    if x <= 3/6:
        return 3 
    if x <= 4/6:
        return 4
    if x<= 5/6 :
        return 5
    if x<= 6/6:
        return 6
```

```python
X['score_SANTE'] = X['SCORE_SYNT'].apply(Score_sante)
```
